```python
drop_cols = [
    "is_canceled",
    "reservation_status",
    "reservation_status_date",
    "assigned_room_type",
    "room_changed"
]

X = df_master.drop(columns=drop_cols)
y = df_master["is_canceled"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Treat agent and company as categorical IDs

categorical_cols = [
    "hotel",
    "arrival_date_month",
    "meal",
    "country",
    "market_segment",
    "distribution_channel",
    "reserved_room_type",
    "deposit_type",
    "customer_type",
    "booking_source_type",
    "waiting_list_group",
    "adr_group",
    "adr_category",
    "special_request_category",
    "agent",
    "company"
]

print("Number of categorical features:", len(categorical_cols))
```

```python
for col in categorical_cols:
    print(
        f"{col:30} "
        f"unique={df_master[col].nunique(dropna=True):5} "
        f"missing={df_master[col].isna().sum():5}"
    )
```

```python
# Columns that should NOT be used as predictors

drop_cols = [
    "is_canceled",              # Target
    "reservation_status",       # Leakage
    "reservation_status_date",  # Leakage
    "assigned_room_type",      # Potential post-booking information
    "room_changed"              # Derived from room assignment change
]

X = df_master.drop(columns=drop_cols)
y = df_master["is_canceled"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numeric_features = X.select_dtypes(
    include=[np.number]
).columns.tolist()

# Agent and company are IDs, so treat them as categorical
categorical_features.extend(["agent", "company"])

# Remove them from numeric features
numeric_features = [
    col for col in numeric_features
    if col not in ["agent", "company"]
]

print("Categorical features:", len(categorical_features))
print(categorical_features)

print("\nNumeric features:", len(numeric_features))
print(numeric_features)
```

```python
X_model = X.copy()

# Categorical columns
for col in categorical_features:
    X_model[col] = (
        X_model[col]
        .fillna("Unknown")
        .astype(str)
    )

# Numerical columns
for col in numeric_features:
    X_model[col] = X_model[col].fillna(
        X_model[col].median()
    )

print("Remaining missing values:",
      X_model.isna().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_model,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        ),
        (
            "numeric",
            "passthrough",
            numeric_features
        )
    ]
)
```

```python
X_train_encoded = preprocessor.fit_transform(X_train)

X_test_encoded = preprocessor.transform(X_test)

print("Encoded training shape:", X_train_encoded.shape)
print("Encoded test shape:", X_test_encoded.shape)
```

```python
from lightgbm import LGBMClassifier

lgbm_model = LGBMClassifier(
    objective="binary",
    random_state=42,
    n_jobs=-1
)

lgbm_model.fit(
    X_train_encoded,
    y_train
)

print("LightGBM training completed.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

lgbm_pred = lgbm_model.predict(X_test_encoded)
lgbm_prob = lgbm_model.predict_proba(X_test_encoded)[:, 1]

print("LightGBM Baseline Results")
print("-" * 30)
print("Accuracy :", round(accuracy_score(y_test, lgbm_pred), 4))
print("Precision:", round(precision_score(y_test, lgbm_pred), 4))
print("Recall   :", round(recall_score(y_test, lgbm_pred), 4))
print("F1 Score :", round(f1_score(y_test, lgbm_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, lgbm_prob), 4))
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train_encoded,
    y_train
)

print("XGBoost training completed.")
```

```python
xgb_pred = xgb_model.predict(X_test_encoded)
xgb_prob = xgb_model.predict_proba(X_test_encoded)[:, 1]

print("XGBoost Baseline Results")
print("-" * 30)
print("Accuracy :", round(accuracy_score(y_test, xgb_pred), 4))
print("Precision:", round(precision_score(y_test, xgb_pred), 4))
print("Recall   :", round(recall_score(y_test, xgb_pred), 4))
print("F1 Score :", round(f1_score(y_test, xgb_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, xgb_prob), 4))
```

```python
from sklearn.model_selection import RandomizedSearchCV

lgbm_params = {
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators": [100, 200, 500],
    "max_depth": [4, 6, 8],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]
}

lgbm_base = LGBMClassifier(
    objective="binary",
    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

lgbm_search = RandomizedSearchCV(
    estimator=lgbm_base,
    param_distributions=lgbm_params,
    n_iter=20,
    scoring="roc_auc",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

lgbm_search.fit(X_train_encoded, y_train)

print("Best parameters:")
print(lgbm_search.best_params_)

print("\nBest CV ROC-AUC:")
print(round(lgbm_search.best_score_, 4))
```

```python
best_lgbm = lgbm_search.best_estimator_

lgbm_tuned_pred = best_lgbm.predict(X_test_encoded)
lgbm_tuned_prob = best_lgbm.predict_proba(
    X_test_encoded
)[:, 1]

print("Tuned LightGBM Results")
print("-" * 30)

print("Accuracy :",
      round(accuracy_score(y_test, lgbm_tuned_pred), 4))

print("Precision:",
      round(precision_score(y_test, lgbm_tuned_pred), 4))

print("Recall   :",
      round(recall_score(y_test, lgbm_tuned_pred), 4))

print("F1 Score :",
      round(f1_score(y_test, lgbm_tuned_pred), 4))

print("ROC-AUC  :",
      round(roc_auc_score(y_test, lgbm_tuned_prob), 4))
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV

xgb_params = {
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators": [100, 200, 500],
    "max_depth": [4, 6, 8],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]
}

xgb_base = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_search = RandomizedSearchCV(
    estimator=xgb_base,
    param_distributions=xgb_params,
    n_iter=20,
    scoring="roc_auc",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

xgb_search.fit(X_train_encoded, y_train)

print("Best XGBoost parameters:")
print(xgb_search.best_params_)

print("\nBest CV ROC-AUC:")
print(round(xgb_search.best_score_, 4))
```

```python
best_xgb = xgb_search.best_estimator_

xgb_tuned_pred = best_xgb.predict(X_test_encoded)
xgb_tuned_prob = best_xgb.predict_proba(X_test_encoded)[:, 1]

print("Tuned XGBoost Results")
print("-" * 30)

print("Accuracy :", round(accuracy_score(y_test, xgb_tuned_pred), 4))
print("Precision:", round(precision_score(y_test, xgb_tuned_pred), 4))
print("Recall   :", round(recall_score(y_test, xgb_tuned_pred), 4))
print("F1 Score :", round(f1_score(y_test, xgb_tuned_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, xgb_tuned_prob), 4))
```

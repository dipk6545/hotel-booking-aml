```python
# Create binary domestic traveler flag
df['is_domestic'] = (df['country'] == 'PRT').astype(int)
```

```python
# 1. Map month names to numbers (1 - 12)
month_map = {
    'January': 1, 'February': 2, 'March': 3, 'April': 4,
    'May': 5, 'June': 6, 'July': 7, 'August': 8,
    'September': 9, 'October': 10, 'November': 11, 'December': 12
}
month_num = df['arrival_date_month'].map(month_map)

# 2. Apply Sin & Cos Cyclical Encoding
df['arrival_month_sin'] = np.sin(2 * np.pi * month_num / 12)
df['arrival_month_cos'] = np.cos(2 * np.pi * month_num / 12)
```

```python
df['total_stay_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['is_weekend_only'] = (
    (df['stays_in_week_nights'] == 0) & (df['stays_in_weekend_nights'] > 0)
).astype(int)
```

```python
# 1. Total party size
df['total_guests'] = df['adults'] + df['children'].fillna(0) + df['babies']

# 2. Family flag (strong commitment signal)
df['has_babies'] = (df['babies'] > 0).astype(int)
df['is_family'] = ((df['children'] > 0) | (df['babies'] > 0)).astype(int)
```

# CAT-BOOST

```python
df_tree = df[tree_df_cols].copy()
```

```python
# Fill numeric columns with median, and text columns with mode
for col in df_tree.columns:
    if df_tree[col].dtype in ['int64', 'float64']:
        df_tree[col].fillna(df_tree[col].median(), inplace=True)
    else:
        df_tree[col].fillna(df_tree[col].mode()[0], inplace=True)

print("Null count:", df_tree.isnull().sum().sum())
```

```python
print("Tree DataFrame Shape:", df_tree.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, RocCurveDisplay
from catboost import CatBoostClassifier
import matplotlib.pyplot as plt

X = df_tree.drop(columns=['is_canceled'])
y = df_tree['is_canceled']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

print(f"X_train: {X_train.shape} | X_test: {X_test.shape}")
print(f"Train class balance:\n{y_train.value_counts(normalize=True)}")
```

```python
# 1. Define categorical column names
cat_features = [
    'hotel', 
    'arrival_date_month', 
    'meal', 
    'market_segment', 
    'deposit_type', 
    'customer_type', 
    'booking_source_type'
]

# 2. Ensure string data type
for col in cat_features:
    X_train[col] = X_train[col].astype(str)
    X_test[col] = X_test[col].astype(str)
```

```python
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
from sklearn.metrics import (
    roc_auc_score, 
    classification_report, 
    precision_recall_curve, 
    f1_score, 
    accuracy_score,
    precision_score,   # ◄ Added
    recall_score       # ◄ Added
)

# 1. Initialize Model
final_cat_model = CatBoostClassifier(
    depth=10,
    learning_rate=0.05,
    l2_leaf_reg=7,
    iterations=1000,
    cat_features=cat_features,
    eval_metric='AUC',
    loss_function='Logloss',
    task_type='GPU',
    random_seed=42,
    verbose=0,
)

# 2. Fit Model
final_cat_model.fit(
    X_train, y_train,
    eval_set=(X_test, y_test),
    early_stopping_rounds=50,
    verbose=0
)

# 3. Predict Probabilities & Compute ROC-AUC
y_pred_proba = final_cat_model.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, y_pred_proba)

# 4. Find Best Probability Threshold (Maximizing F1-Score)
precisions, recalls, thresholds = precision_recall_curve(y_test, y_pred_proba)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)
best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]
best_f1 = f1_scores[best_idx]

# 5. Predictions at Default (0.5) vs Optimal Threshold
y_pred_default = (y_pred_proba >= 0.5).astype(int)
y_pred_optimal = (y_pred_proba >= best_threshold).astype(int)

# 6. Summary Comparison DataFrame (Default vs Optimal Threshold)
summary_df = pd.DataFrame([
    {
        'Model': 'CatBoost',
        'Threshold': 0.50,
        'ROC-AUC': round(roc_auc, 4),
        'Accuracy': round(accuracy_score(y_test, y_pred_default), 4),
        'Precision': round(precision_score(y_test, y_pred_default), 4),
        'Recall': round(recall_score(y_test, y_pred_default), 4),
        'F1-Score': round(f1_score(y_test, y_pred_default), 4)
    },
    {
        'Model': 'CatBoost (Tuned Threshold)',
        'Threshold': round(best_threshold, 4),
        'ROC-AUC': round(roc_auc, 4),
        'Accuracy': round(accuracy_score(y_test, y_pred_optimal), 4),
        'Precision': round(precision_score(y_test, y_pred_optimal), 4),
        'Recall': round(recall_score(y_test, y_pred_optimal), 4),
        'F1-Score': round(f1_score(y_test, y_pred_optimal), 4)
    }
])

# 7. Detailed Classification Report DataFrame (Optimal Threshold)
report_dict = classification_report(y_test, y_pred_optimal, output_dict=True)
detailed_report_df = pd.DataFrame(report_dict).transpose().round(4)

# Display DataFrames
print("=== Summary Performance Table ===")
display(summary_df)

print("\n=== Detailed Classification Report (Optimal Threshold) ===")
display(detailed_report_df)
```
